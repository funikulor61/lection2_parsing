```python
def clean_links(links: list):
    print(f'Start size of links list: {len(links)}')
    clear_links = []
    very_clear_links = []

    for item in links:
        if item == None or 'forum.' in item or 'tech.' in item:
            pass
        else:
            clear_links.append(item)
    print(f'Current size of links list: {len(clear_links)}')

    for item in clear_links:
        if 'prices' in item or 'reviews' in item or 'used' in item:
            pass
        else:
            very_clear_links.append(item)
    print(f'Finish size of links list: {len(very_clear_links)}')

    return very_clear_links
    
```

```python
def get_character():
    slovli = {}
    slovli.update({'Название ' :driver.find_element(By.CLASS_NAME, value='catalog-masthead__title').text})
    # 1. Находим все строки (tr) внутри таблиц характеристик
    rows = driver.find_elements(By.CSS_SELECTOR, ".product-specs__table tr")
    
    # 2. Перебираем их построчно
    for row in rows:
        tds = row.find_elements(By.TAG_NAME, "td")
        
        # Если в строке ровно 2 ячейки — это характеристика и её значение
        if len(tds) == 2:
            if not tds[1].text:
                if tds[1].find_elements(By.CLASS_NAME, value='i-x'):
                    slovli.update({tds[0].text: 'No'})
                else :
                    slovli.update({tds[0].text: 'Yes'})
            else:
                slovli.update({tds[0].text:tds[1].text})
                
    # Выводим красивый результат
    print(f"Успешно собрано характеристик: {len(slovli)}")
    with open('harakteristiki.txt', 'a', encoding='utf-8') as file:
        for key, value in slovli.items():
            file.write(f"{key}: {value}\n")
    for key,value in slovli.items():
        print(f'{key}  ::-_-_-:: {value}')

    return slovli

get_character()

```
